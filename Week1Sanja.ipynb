```python
print("Welcome to GPA Calculator")
print("Enter GPA one by one and press enter at the end")
points = {
    "A+": 4.0,
    "A":4.0,
    "A-":3.67,
    "B+":3.33,
    "B":3.0,
    "B-":2.7,
    "C+": 2.33,
    "C":2.00,
    "D+":1.33,
    "D":1.00,
    "F":0
}
no_of_course = 0
total_point=0
done = False
while not done:
    grade = input()
    if grade == "":
        done = True
    elif grade not in points:
        print("Not a valid grade")
    else:
        no_of_course+=1
        total_point = total_point + points[grade]
if no_of_course > 0:
        print("Your GPA is :", total_point/no_of_course)
    
```
