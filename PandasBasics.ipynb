```python
#Get all the rows where milage is >=30

con = data_csv['Milage']>=30
```

```python
#Filter the data where milage>=30 and location is mumbai
con1 = data_csv["Milage"]>=30
con2 = data_csv["Location"] == "Mumbai"
data_csv[con1 & con2]
```

```python
#Filter the data where milage>=30 or location is mumbai
```

```python
con1 = data_csv["Milage"]>=30
con2 = data_csv["Location"] == "Mumbai"
data_csv[con1 | con2]
```

```python
#not 
#Filter all the rows where not condition for milage >=30 and location is Mumbai

con1 = data_csv["Milage"]>=30
con2 = data_csv["Location"] == "Mumbai"
data_csv[~(con1 & con2)]
```
